```python
import pandas as pd
import numpy as np
from sklearn import linear_model
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("walking_routes_cleaned.csv")
```

```python
df
```

```python
# X must be 2D (DataFrame), y is 1D (Series)
X = df[['google_distance_km']]
y = df['google_walking_time_min']
```

```python
# Create model
model = linear_model.LinearRegression()
```

```python
# Train model
model.fit(X, y)
```

```python
new_distance = [[3.5]]

predicted_time = model.predict(new_distance)

print("Predicted walking time:", predicted_time[0])
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, label='Actual observations')

plt.plot(X, model.predict(X),
         label='Regression line')

plt.xlabel('google_distance_km')
plt.ylabel('google_walking_time_min')

plt.title('Linear Regression: Walking Time Prediction')

plt.legend()
plt.show()
```
